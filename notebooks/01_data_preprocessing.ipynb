```python
# %% [markdown]
# # Step 1: Data Generation & Preprocessing Pipeline
# This notebook handles creating a realistic diesel repair dataset and preparing it for ML modeling.

# %%
import os
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler

print("🚀 Initializing Vehicle AI Data Pipeline...")

# Ensure data directory exists
os.makedirs('../data', exist_ok=True)

# %% [markdown]
# ### 1. Simulate the Domain-Specific Diesel Workshop Dataset
# Since real-world maintenance logs are hard to access, we generate 1,000 realistic 
# job cards capturing common diesel mechanical issues.

# %%
np.random.seed(42)
data_size = 1000

# Generating realistic vehicle attributes
mileage = np.random.randint(20000, 350000, size=data_size)
vehicle_age = np.random.randint(1, 15, size=data_size)
fuel_type = ['Diesel'] * data_size  

# Real diesel repair shop scenarios mapping (Symptom -> Fault -> Fix -> Base Cost)
symptoms_pool = [
    ("Black smoke from exhaust, severe power loss under heavy load", "Turbocharger Fault", "Replace Turbocharger & Intercooler", 1200.00),
    ("Engine cranks but struggles to start cold, rough idle on startup", "Glow Plug / Fuel System", "Replace Glow Plugs & Fuel Filter", 350.00),
    ("Knocking noise from engine bay, poor fuel economy, engine misfiring", "Fuel Injector Failure", "Calibrate and Replace Diesel Injectors", 950.00),
    ("Excessive blue smoke from tailpipe, high engine oil consumption", "Piston Ring / Valve Seal Wear", "Engine Top-Overhaul", 2200.00),
    ("Battery warning light on dashboard, steering heavy, voltage dropping", "Alternator / Drive Belt", "Replace Alternator & Serpentine Belt", 450.00)
]

chosen_scenarios = [symptoms_pool[np.random.choice(len(symptoms_pool))] for _ in range(data_size)]

# Construct DataFrame
df = pd.DataFrame({
    'mileage': mileage,
    'vehicle_age': vehicle_age,
    'symptom_text': [s[0] for s in chosen_scenarios],
    'fault_category': [s[1] for s in chosen_scenarios],
    'repair_action': [s[2] for s in chosen_scenarios],
    'estimated_cost': [s[3] for s in chosen_scenarios]
})

# Add variance to cost based on mileage and random distribution to simulate real pricing
df['estimated_cost'] = df['estimated_cost'] + (df['mileage'] * 0.0015) + np.random.normal(0, 75, data_size)
df['estimated_cost'] = df['estimated_cost'].round(2)

# Save the raw data locally
df.to_csv('../data/raw_workshop_data.csv', index=False)
print("💾 Raw dataset created and saved to '../data/raw_workshop_data.csv'")

# %% [markdown]
# ### 2. Feature Encoding & Preprocessing

# %%
# Encode categorical targets for classification tasks
label_encoder_fault = LabelEncoder()
df['fault_category_encoded'] = label_encoder_fault.fit_transform(df['fault_category'])

# Scale numerical metrics for algorithms like XGBoost/Neural Networks
scaler = StandardScaler()
df[['mileage_scaled', 'vehicle_age_scaled']] = scaler.fit_transform(df[['mileage', 'vehicle_age']])

print("\n🔢 Class mapping for Fault Category:")
for index, label in enumerate(label_encoder_fault.classes_):
    print(f"  {index} -> {label}")

# %% [markdown]
# ### 3. Train-Test Split

# %%
X = df[['mileage_scaled', 'vehicle_age_scaled', 'symptom_text']]
y_fault = df['fault_category_encoded']
y_cost = df['estimated_cost']

# Train-test split (80% training, 20% validation)
X_train, X_test, y_train_fault, y_test_fault = train_test_split(X, y_fault, test_size=0.2, random_state=42)
_, _, y_train_cost, y_test_cost = train_test_split(X, y_cost, test_size=0.2, random_state=42)

print(f"\n📂 Splits Created Successfully:")
print(f"   Train set size: {X_train.shape[0]} samples")
print(f"   Test set size: {X_test.shape[0]} samples")
```
